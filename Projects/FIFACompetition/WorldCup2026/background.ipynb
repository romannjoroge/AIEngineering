# Background Pipeline

This pipeline takes in an Excel file and gets player statistics from the internet from players in a sheet. These statistics are then stored and used in prediction pipeline

```python
import math, time
import pandas as pd
from datetime import datetime
from helpers.background import Player
```

```python
def get_player_statistics(filename: str, sheet_name: str):
    """ 
    File to get statistics of player from excel sheet from the internet and store them locally for later use
    
    Args:
        filename (scalar): Name of the excel file
        sheet_name (scalar): Name of the sheet with player information
    
    """
    players = pd.read_excel(filename, sheet_name=sheet_name)
    
    batch_size = 100
    total_batches = math.ceil(players.shape[0] / batch_size)
    print(f"Total batches: {total_batches}")
    extracted_players = []
    problematic_players = []

    file_name = f"data/player_statistics_{datetime.now()}.csv"
        
    with open(file_name, 'a') as file:
        # Insert csv file header
        file.write("id,appearances,team,goals,from,to\n")
        
        # Get players
        for batch_num in range(total_batches):
            # Getting indexes of batch
            print(f"Processing batch #{batch_num}")
            start = batch_num * batch_size
            if batch_num == total_batches - 1:
                end = players.shape[0]
            else:
                end = start + batch_size
            
            # Storing details of players in batch 
            df_segment = players.iloc[start:end]
            for row in df_segment.itertuples():                  
                player = Player(
                    id=row.ID,
                    wikipedia_url=row.wikipedia,
                    problematic_players=problematic_players
                )
                
                extracted_players.append(player)
                # Insert entries in CSV
                for entry in player.to_csv():
                    file.write(entry)
                
            time.sleep(10)
        
```

```python
filename = "data/World Cup 2026.xlsx"
sheet_name = "Players"
get_player_statistics(filename=filename, sheet_name=sheet_name)
```
